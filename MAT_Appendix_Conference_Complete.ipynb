# MAT-Appendix Conference-Complete Run
Confirmed appendicitis-only severity cohort. Runs five classical baselines, FT-Transformer, missingness ablation and MAT-Appendix on identical five OOF folds; produces complete metrics, calibration, XAI, subgroup analyses, PKL and GitHub paper assets.


```python
import requests
BASE='https://raw.githubusercontent.com/AzizulHakim00/MAT-Appendix/main/src/v2'
part_names=[f'part_{i:02d}.pyfrag' for i in range(1,7)]+['part_07_pre.pyfrag','part_07b.pyfrag','part_07_post.pyfrag','part_08.pyfrag']
parts=[]
for name in part_names:
    url=f'{BASE}/{name}'
    response=requests.get(url,timeout=60)
    response.raise_for_status()
    parts.append(response.text)
source=''.join(parts)
compile(source,'MAT_Appendix_Conference_Complete','exec')
exec(source,globals())
```
