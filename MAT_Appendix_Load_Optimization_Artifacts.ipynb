# Load Latest MAT-Appendix Optimization Artifacts
Loads saved PKL without retraining.


```python
import os, json, joblib, pandas as pd
from pathlib import Path
try:
    from google.colab import drive
    drive.mount('/content/drive', force_remount=False)
    root = Path('/content/drive/MyDrive/MAT-Appendix/optimization_runs')
except Exception:
    root = Path('/content/MAT-Appendix/optimization_runs')
runs = sorted([p for p in root.glob('*') if p.is_dir()])
if not runs:
    raise FileNotFoundError(f'No optimization run found under {root}')
latest = runs[-1]
bundle_path = latest/'mat_appendix_optimized_reproducibility.pkl'
bundle = joblib.load(bundle_path)
print('Loaded:', bundle_path)
print('Run:', bundle['run_id'])
metrics = pd.DataFrame(bundle['metrics'])
target_gap = pd.DataFrame(bundle['target_gap'])
paired = pd.DataFrame(bundle['paired_bootstrap'])
display(metrics)
display(target_gap[target_gap['Model']=='Optimized Stacked Ensemble'])
display(paired)
y = bundle['predictions']['y']
oof_probability = bundle['predictions']['oof_probability']
oof_class = bundle['predictions']['oof_class']
```
