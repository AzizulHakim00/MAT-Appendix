# Load Latest Conference-Complete MAT-Appendix Artifacts
Loads the latest GitHub PKL without retraining any model.


```python
import sys, subprocess, requests, joblib, pandas as pd
subprocess.check_call([sys.executable,'-m','pip','install','-q','joblib'])
BASE='https://raw.githubusercontent.com/AzizulHakim00/MAT-Appendix/main/results'
run_id=requests.get(f'{BASE}/LATEST_RUN.txt',timeout=30).text.strip()
if not run_id:
    raise RuntimeError('LATEST_RUN.txt is empty. Ensure the conference run was pushed from Colab.')
url=f'{BASE}/runs/{run_id}/mat_appendix_complete_reproducibility.pkl'
response=requests.get(url,timeout=180)
response.raise_for_status()
path='/content/mat_appendix_complete_reproducibility.pkl'
open(path,'wb').write(response.content)
bundle=joblib.load(path)
metrics=pd.DataFrame(bundle['metrics']['primary'])
operating_points=pd.DataFrame(bundle['metrics']['operating_points'])
paired_bootstrap=pd.DataFrame(bundle['metrics']['paired_bootstrap'])
subgroups=pd.DataFrame(bundle['subgroups'])
print('Loaded run:',bundle['run_id'])
print('Cohort:',bundle['dataset']['cohort_definition'])
display(metrics)
display(operating_points)
display(paired_bootstrap)
display(subgroups)
print('XAI:',bundle['xai']['status'])
print('Available keys:',bundle.keys())
```
