# MAT-Appendix Multimodal Dataset Auto-Download
Downloads the official checksum-verified Regensburg ultrasound + tabular dataset from Zenodo and the 2025 Düsseldorf external-validation data from the authors' GitHub repository. No manual upload or Kaggle token is required.

**Date warning:** the repository was released/updated in 2023–2024, but the clinical data were collected earlier. Do not describe this as a 2023–2026 patient cohort.


```python
!pip -q install pandas openpyxl pillow tqdm requests
import urllib.request

SOURCE_COMMIT = "bf5ea6437a6c0aeaf4c9b6887c5eb8aac94d259b"
SOURCE_URL = (
    "https://raw.githubusercontent.com/AzizulHakim00/MAT-Appendix/"
    f"{SOURCE_COMMIT}/src/data/multimodal_data_auto_download.py"
)
print("Loading pinned multimodal data loader...")
print("Source commit:", SOURCE_COMMIT)
source = urllib.request.urlopen(SOURCE_URL).read().decode("utf-8")
exec(compile(source, SOURCE_URL, "exec"), globals(), globals())
```
